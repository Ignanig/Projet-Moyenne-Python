```python
#initialisation de mon dictionnaire et de mes deux variables
#les notes sont en clés et les coeff sont la valeur de chaque clé

note_coeff = {14:10,12:10,16:7,10:7,0:0,20:5,0:0,12:6,0:0,10:5,11:4,14:2,11:5,5:5,0:0,12:20,12:20,0:0,0:0,0:0,12:7}
somnote = 0
somcoeff = 0

#je parcours chaque couple dans mon dictionnaire pour avoir la note et le coeff que je stocke dans mes variables
for cle, item in note_coeff.items():
    somnote += cle * item
    somcoeff += item

#je calcule la moyenne en divisant la somme des note par la somme des coeff
moyenne = somnote / somcoeff
#j'affiche la moyenne
print ("Votre moyenne est",moyenne,"/20")
```
