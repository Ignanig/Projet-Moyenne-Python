```python
#initialisation de mes deux listes et de mes deux variables

note = (12,10,12.2,15,9,14,10,12,11,17,13,15,2,8,12,14,13,11,9,7,5)
coeff = (10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7)
somnote = 0
somcoeff = 0

#boucle for pour parcourir la longueur de liste note et incrémenter les valeurs a chaque tour de boucle
for i in range(len(note)):
    somnote += note[i] * coeff[i]
    somcoeff += coeff[i]
#calcul de la moyenne
moyenne = somnote / somcoeff
#affichage de la moyenne
print("Votre moyenne est",moyenne,"/20")
```
